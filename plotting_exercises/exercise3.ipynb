```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
m_e = 9.109e-31
e   = 1.602e-19
B   = 1e-4            # <-- was e-4
v   = 5.0
v_z = 3.0
```

```python
omega = e * B / m_e             # cyclotron frequency
r     = m_e * v / (e * B)       # gyroradius

T = 2 * np.pi / omega           # one period
ts = np.linspace(0, 2 * T, 500) # <-- a few orbits, dense sampling
```

```python
xs =  r * np.sin(omega * ts)    # <-- proper parametric circle
ys =  r * np.cos(omega * ts)
zs =  v_z * ts
```

```python
fig = plt.figure()

ax = fig.add_subplot(111, projection='3d')

ax.plot(xs, ys, zs, linewidth=1.5, color='blue')

ax.set_box_aspect((10, 10, 10))
# ax.view_init(elev=-50, azim=30, roll = 90)
ax.set_xlabel('x-axis [m]')
ax.set_ylabel('y-axis [m]')
ax.set_zlabel('z-axis [m]')
ax.set_title('Electron Helix in B-field')
ax.xaxis.pane.fill = True
ax.grid(False)

plt.show()
```
