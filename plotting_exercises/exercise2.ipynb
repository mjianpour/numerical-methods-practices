```python
from astropy.io import fits
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import LogNorm
```

```python
image = fits.open("phys_src_01388.fits")
```

```python
z = np.random.rand(100, 100)
```

```python
z
```

```python
image.info()
```

```python
sci = image["SCI"].data
```

```python
fig, ax = plt.subplots()
ax = plt.imshow(z, cmap="inferno", vmax = 0.5)
plt.colorbar()
plt.show()
```

```python
plt.imshow(sci, origin='lower', cmap='gray', vmin=sci.min() * 2, vmax=sci.max()*1.05, interpolation='bilinear')
```

```python
fig, axd = plt.subplots(1,4, figsize=(17,17))


axd[0].imshow(sci, origin="lower", cmap="bone", vmin = sci.min() * 2, vmax = sci.max())
axd[1].imshow(sci, origin="lower", cmap="inferno", vmin = sci.min() * 2, vmax = sci.max()),
axd[2].imshow(sci, origin="lower", vmin = sci.min() * 2, vmax = sci.max())
axd[3].imshow(sci, origin="lower", cmap="twilight", interpolation='bilinear', vmin = sci.min() * 2, vmax = sci.max())
# ax = plt.imshow(sci, origin="lower", extent=[0, 100, 0, 50])
plt.show()
```
