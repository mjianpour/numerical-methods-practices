```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
```

```python
# 1. Setup figure and initial artist
fig, ax = plt.subplots()
x = np.linspace(0, 2 * np.pi, 200)
y = np.sin(x)
line, = ax.plot([], [], lw=2) # Start with empty data
ax.set_xlim(0, 2 * np.pi)
ax.set_ylim(-1.2, 1.2)

# 2. Define update function
def update(frame):
    # frame is the current index from the generator
    line.set_data(x[:frame], y[:frame])
    return line,

# 3. Create the animation
# frames is a generator yielding 0, 1, 2... up to len(x)
ani = animation.FuncAnimation(
    fig, 
    update, 
    frames=len(x), 
    interval=20, 
    blit=True # blit=True improves performance by redrawing only changed parts
)

plt.show()
```

```python
ani.save('sine_wave.gif', writer='pillow', fps=30)
```
