```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# defining parameters
```

```python
xs = np.linspace(-1, 1, 300)
α = 5
ys = [np.exp(-α * x**2) for x in xs]
ys_n = (np.sqrt(np.pi / α)) * np.array([np.exp(-α * x**2) for x in xs])

# simplest way to plot a set of x and y
plt.plot(xs, ys)
```

```python
# Let's make it fancier and more professional!
```

```python
# It's time to take the full control over the plot! 
```

```python
from matplotlib.ticker import MultipleLocator
```

```python
# Common String Values: 
# 'upper right' (default)
# 'upper left'
# 'lower left'
# 'lower right'
# 'center left'
# 'center right'
# 'lower center'
# 'upper center'
# 'center'
```

```python
fig, ax = plt.subplots(
    nrows=1, 
    ncols=2,
    sharey=True,
    figsize = (8,4),    # dimensions of all plots together.  
    # width_ratios=[4, 1], # height_ratios=[] for column
    layout='constrained' # Prevents overlapping titles/labels automatically
)
ax[0].plot(xs, ys, color="purple", linewidth=5)
ax[0].plot(xs, ys_n, color="red", linewidth=1, linestyle="dashed")
ax[0].set_title("Unnormalized")
ax[0].grid()
ax[0].legend(labels = ["Unnormalized", "Normalized"], loc='upper right')
ax[0].set_ylabel("y")
ax[0].set_xlabel("x")
ax[0].yaxis.set_minor_locator(MultipleLocator(0.05))

ax[1].plot(xs, ys_n, color="purple", linewidth=1)
ax[1].set_title("Normalized")
ax[1].set_xlabel("x")
ax[1].grid()

plt.show()
```
