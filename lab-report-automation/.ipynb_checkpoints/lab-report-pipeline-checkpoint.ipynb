```python
import re
from langchain.agents.middleware import AgentMiddleware, hook_config
from langchain_core.messages import HumanMessage, AIMessage
from typing import Any

class LabReportValidatorMiddleware(AgentMiddleware):
    def __init__(self, max_retries: int = 2):
        super().__init__()
        self.max_retries = max_retries

    def _check_significant_figures(self, text: str) -> list[str]:
        """
        Finds all patterns like '3.457 ± 0.0124' or '3.46±0.01' and checks 
        if the uncertainty has exactly 1 significant figure, and if the 
        value's decimal places match the uncertainty's decimal places.
        """
        issues = []
        # Regex to find: [number] ± [number] (allows spaces)
        # Matches things like: 3.457 ± 0.0124, 12.5±0.2, 3 ± 1
        pattern = r"([+-]?\d+(?:\.\d+)?)\s*[±]\s*([+-]?\d+(?:\.\d+)?)"
        
        matches = re.finditer(pattern, text)
        
        for match in matches:
            val_str = match.group(1)
            unc_str = match.group(2)
            
            # 1. Check if uncertainty has exactly 1 significant figure
            # Remove decimals and leading zeros to find the sig fig count
            unc_clean = unc_str.lstrip('0.').lstrip('0')
            if not unc_clean:
                continue # Skip if it's 0.00
                
            if len(unc_clean) != 1:
                issues.append(
                    f"Format error: `{match.group(0)}`. The uncertainty (`{unc_str}`) "
                    f"has more than 1 significant figure. It must be rounded to 1 sig fig (e.g., 0.0124 -> 0.01)."
                )
                continue

            # 2. Check if decimal places match
            # Get decimal places of uncertainty
            if '.' in unc_str:
                unc_decimals = len(unc_str.split('.')[1])
            else:
                unc_decimals = 0
                
            # Get decimal places of value
            if '.' in val_str:
                val_decimals = len(val_str.split('.')[1])
            else:
                val_decimals = 0

            if val_decimals != unc_decimals:
                issues.append(
                    f"Format error: `{match.group(0)}`. The value (`{val_str}`) has {val_decimals} decimal places, "
                    f"but the uncertainty (`{unc_str}`) has {unc_decimals}. "
                    f"The value must be rounded to match the uncertainty (e.g., 3.457 ± 0.01 -> 3.46 ± 0.01)."
                )
                
        return issues

    @hook_config(can_jump_to=["model"])
    def after_model(self, state: dict[str, Any], runtime: Any) -> dict[str, Any] | None:
        messages = state["messages"]
        last_msg = messages[-1]

        if not isinstance(last_msg, AIMessage) or last_msg.tool_calls:
            return None

        content = last_msg.content or ""
        retry_count = state.get("_validation_retry", 0)
        
        issues = []

        # --- Rule 1: TODO Placeholders ---
        if "====" not in content:
            issues.append("Response does not contain any `====` placeholder markers for manual image/chart insertion.")

        # --- Rule 2: Significant Figures & Error Reporting ---
        math_issues = self._check_significant_figures(content)
        issues.extend(math_issues)

        # --- Trigger Retry if issues found ---
        if issues and retry_count < self.max_retries:
            state["_validation_retry"] = retry_count + 1
            
            # Build the correction prompt
            correction_text = (
                "Your previous response violates the lab report formatting rules. Specific issues found:\n\n"
                + "\n".join(f"- {i}" for i in issues)
                + "\n\nPlease rewrite the entire output. Remember the strict rule for numerical reporting:\n"
                  "1. The uncertainty must be rounded to exactly ONE significant figure.\n"
                  "2. The measured value must be rounded to the SAME number of decimal places as the uncertainty.\n"
                  "Example: If you have 3.457 ± 0.0124, it MUST be written as 3.46 ± 0.01."
            )
            
            messages.append(HumanMessage(content=correction_text))
            return {"messages": messages, "jump_to": "model"}

        # Clean up on success
        if "_validation_retry" in state:
            del state["_validation_retry"]
            
        return None
```

```python
from langchain.agents.middleware import ModelRetryMiddleware, ModelCallLimitMiddleware

agent = create_agent(
    model=ChatDeepSeek(model = "deepseek-chat", temperature=0.7),
    tools=tools,
    middleware=[
        ModelCallLimitMiddleware(run_limit=10),      # Prevent infinite loops
        ModelRetryMiddleware(max_retries=2),         # Network-level retry
        LabReportValidatorMiddleware(max_retries=2), # Content quality check
    ],
)
```

```python
result = agent.invoke({
    "messages": [{"role": "user", "content": todo_prompt}]
})
print(result["messages"][-1].content)
```
